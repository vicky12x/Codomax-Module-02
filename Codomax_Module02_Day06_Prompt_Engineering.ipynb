```python
# Zero-shot Prompting

sentence = "I love learning Python programming."

prompt = f"""
Classify the following sentence
as Positive, Negative or Neutral.

Sentence: {sentence}

Give only the classification.
"""

print(prompt)
```

```python
# Few-shot Prompting

prompt = """
Classify sentences as Positive or Negative.

Examples:

I love this product. = Positive

This product is terrible. = Negative

The service is excellent. = Positive

Now classify:

The mobile phone is very bad.
"""

print(prompt)
```

```python
# Reusable Prompt Template

def create_prompt(role, topic, level):

    prompt = f"""
    Act as a {role}.

    Explain {topic}
    for a {level} student.

    Give simple explanations
    and practical examples.
    """

    return prompt


result = create_prompt(
    "Java Teacher",
    "Object Oriented Programming",
    "beginner"
)

print(result)
```
